```python
#Program 1: List creation, indexing and slicing
fruits = ["apple", "banana", "cherry", "mango"]
print(fruits)
print(fruits[0])
print(fruits[-1])
print(fruits[1:3])
```

```python
#Program 2: Common list methods
numbers = [5, 2, 9, 1]
numbers.append(20)
numbers.insert(1, 100)
numbers.sort()
print(numbers)
numbers.remove(100)
print(numbers)
print("Length:", len(numbers))
```

```python
#Program 3: Tuple creation and immutability
point = (10, 20)
print(point[0], point[1])
coordinates = (12.9, 77.6)
lat, lon = coordinates # tuple unpacking
print(lat, lon)
# point[0] = 99 # Uncommenting this line raises TypeError -- tuples are immutable
```

```python
#Program 4: List vs Tuple -- a quick comparison
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
my_list[0] = 99 # allowed
print("List after change:", my_list)
print("Tuple stays same:", my_tuple)
```

```python
#Practice Question 1

numbers = [10, 45, 2, 78, 34, 89, 1]
largest = numbers[0]
smallest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num

print("Largest number:", largest)
print("Smallest number:", smallest)
```
