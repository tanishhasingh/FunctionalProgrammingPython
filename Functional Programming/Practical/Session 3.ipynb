```python
#Program 1: Set creation and operations
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B)
A.add(10)
print("After add:", A)
```

```python
#Program 2: Removing duplicates using a set
numbers = [4, 8, 4, 6, 8, 2, 6, 1]
unique_numbers = list(set(numbers))
print("Original:", numbers)
print("Unique:", sorted(unique_numbers))
```

```python
#Program 3: Dictionary creation and access
student = {"name": "Rhea", "age": 20, "course": "BSc AS&A"}
print(student["name"])
student["age"] = 21
student["college"] = "NMIMS"
print(student)
for key, value in student.items():
    printrint(key, "->", value)
```

```python
#Program 4: Word frequency counter using a dictionary
sentence = "the cat sat on the mat the cat ran"
words = sentence.split()
frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1
    print(frequency)
```

```python
#Practice Question 2

students = { "Aman": 85, "Riya": 92, "Karan": 88, "Sneha": 95, "Rahul": 90}
topper = ""
highest_marks = 0

for name, marks in students.items():
    if marks > highest_marks:
        highest_marks = marks
        topper = name

print("Topper:", topper)
print("Marks:", highest_marks)
```
