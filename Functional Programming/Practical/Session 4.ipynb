```python
#Program 1: String indexing, slicing and basic operations
word = "PYTHON"
print(word[0], word[-1])
print(word[1:4])
print(word[::-1])
print(len(word))
```

```python
#Program 2: Common string methods
name = " Tanisha Singh "
print(name.strip())
print(name.strip().upper())
print(name.strip().lower())
print(name.strip().replace("Tanisha", "Ms. Tanisha"))
print(name.strip().split(" "))
```

```python
#Program 3: Taking input from the user
name = input("Enter your name: ")
age = int(input("Enter your age: "))
print("Hello,", name, "- next year you will be", age + 1)
```

```python
#Program 4: Palindrome check (combining string + input/output)
text = input("Enter a word: ").lower().replace(" ", "")
if text == text[::-1]: 
    print("It IS a palindrome")
else:
    print("It is NOT a palindrome")
```

```python
#Practice Question 1

string = input("Enter a string: ").lower()
vowels = "aeiou"
count = 0

for char in string:
    if char in vowels:
        count += 1

print("Number of vowels:", count)
```
