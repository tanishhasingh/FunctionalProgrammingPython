```python
#SET-18 (10 markers)
#Q1. Write a Python program that accepts a password string from the user and checks whether it is valid according to these rules:
#At least 8 characters long, contains at least one digit, and contains at least one uppercase letter
#Print whether the password is valid, and if not, which rule(s) it fails


password = input("Please enter a password: ") #To input a password 

if(len(password)) < 8: #To check whether length of the password is greater than 8 characters or not
    print("Password is invalid.")
    print("Error : Password must be at least 8 characters long.")
else:
    digit = False
    uppercase = False
for ch in password:
    if ch >= '0' and ch <= '9': #To check whether at least one digit is present in the password
        digit = True
    if ch >= 'A' and ch <= 'Z': #To check whether at least one uppercase letter is present in the password
        uppercase = True

if(digit == False and uppercase == False):
    print("Password is invalid.")
    print("Error : Password does not contain any digits and does not have an uppercase letter.")
elif(digit == False):
    print("Password is invalid.")
    print("Error : Password does not contain any digits.")
elif(uppercase == False):
    print("Password is invalid.")
    print("Error : Password does not contain any uppercase letters.")
else:
    print("Password is valid.")
    
```

```python
#SET-18 (10 markers)
#Q2. A dictionary-based inventory stores each item name as a key and a tuple of (price, quantity) as its value.
#Write a Python program to calculate and display the TOTAL VALUE of the entire inventory (sum of price x quantity for every item).

inventory = { #Making a dictionary with initial values 
    "Pen" : (10,5),
    "Notebook" : (50, 3),
    "Bag" : (500,2)
}

total = 0
for item in inventory:
    price = inventory[item][0] #Obtaining price of each item
    quantity = inventory[item][1] #Obtaining quantity of each item
    total = total + (price*quantity) #Calculating total by multiplying price with quantity and adding it to total

print("Total value of the inventory = ", total) #Printing total value of the inventory
```
